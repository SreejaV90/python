```python
##Find the maximum, minimum, and mean temperatures for the week.

max_temp=np.max(temperatures_w1)
min_temp=np.min(temperatures_w1)
mean_temp=np.mean(temperatures_w1)
print("Max temparature:",max_temp)
print("Min temparature:",min_temp)
print("Mean temparature:",mean_temp)
```

```python
##1. Creating Pandas Series:
import pandas as pd
```

```python
marks=pd.Series(
    data=[95,92,89,85,80],
    index=["Rank1","Rank2","Rank3","Rank4","Rank5"]
)
```

```python
marks
```

```python
#2. Indexing and Slicing:

print("Mark of 1st rank syudent:",marks["Rank1"])
```

```python
#.Marks of the top 3 ranks

top_3=marks.loc["Rank1":"Rank3"]
print("Top 3 rankers are\n",top_3)
```

```python
#1.Mark of the 3rd rank student(USE ILOC)

rank3=marks.iloc[2]
print("Mark of the 3rd rank student:",rank3)
```

```python
#2.Boolean mask - marks > 90

print("Rankers having marks greater than 90:\n",marks[marks>90])
```

```python
##3. Manipulating Series:
##Modify the mark of the student with the 1st rank to 100

marks.loc["Rank1"]=100
print(marks)
```

```python
##Remove the entry corresponding to the last rank from the Series

marks=marks.drop("Rank5")
print(marks)
```

```python
#Compute the CGPA by dividing each mark in the Series by 10.
marks = marks/10
print(marks)
```
