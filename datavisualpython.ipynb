```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
 df = sns.load_dataset("taxis")
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.columns
df.dtypes
```

```python
##Impute missing values
```

```python
df["payment"]=df["payment"].fillna(df["payment"].mode()[0])
df["pickup_borough"]=df["pickup_borough"].fillna(df["pickup_borough"].mode()[0])
df["dropoff_borough"]=df["dropoff_borough"].fillna(df["dropoff_borough"].mode()[0])
```

```python
df
```

```python
df.isnull().sum()
```

```python
#For columns that are critical and cannot be reasonably imputed, remove rows with missing values to maintain data integrity. 
```

```python
##For critical columns like pickup and dropoff datetime, imputation is not reasonable as it would introduce false timestamps. Hence, rows with missing values in these columns were removed using dropna() to maintain data integrity.
```

```python
df=df.drop(columns=["pickup_zone","dropoff_zone"])
```

```python
df.isnull().sum()
```

```python

###Visualizations using Matplotlib/Pandas Plot: 
```

```python
##★LINE CHART 

df["pickup"] = pd.to_datetime(df["pickup"],errors="coerce")
```

```python
df
```

```python
df=df.sort_values(["pickup"])
```

```python
plt.figure(figsize=(15,9))
plt.plot(df["pickup"], df["fare"])
plt.title("Fare Over Time")
plt.xlabel("Pickup Time")
plt.ylabel("Fare")
plt.show()
```

```python
##★ BAR CHART
```

```python
pickup_borough_fare=df.groupby("pickup_borough")["fare"].sum()
pickup_borough_fare
pickup=df.pickup_borough.unique()
pickup
```

```python
plt.figure(figsize=(6,4))
plt.bar(pickup,pickup_borough_fare,color=["blue","red","orange","green"])
plt.title("Total Fare for each Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")
plt.xticks(rotation=45)
plt.show()
```

```python
##PIE CHART

payment_counts=df["payment"].value_counts()
payment_counts
payments=df.payment.unique()
payments
```

```python
plt.pie(
    payment_counts,
    labels=payments,
    autopct="%.2f%%"
)
plt.title("Trip Distribution by payment")
plt.show()
```

### Most of them paid by Cash

```python
## ★ HISTOGRAM


plt.hist(df["distance"],bins=30,color="Brown")
plt.title=("Distribution of trip by Distance")
plt.xlabel("Distance")
plt.ylabel("Number of trips")
plt.show()
```

```python

## Most trips are short distance trip

```

```python
## ★ BOX PLOT
pickup=df.pickup_borough.unique()
pickup
tips=df["tip"]
plt.boxplot(tips,pickup)

plt.xlabel("Pickup Borough")
plt.ylabel("Tip")
plt.show()
```

```python
##Visualizations using Seaborn: 
```

```python
##★ Count Plot 

df = sns.load_dataset("taxis")

sns.countplot(x="pickup_borough", data=df)
plt.title("Number of Trips in Each Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Count of Trips")
plt.show()
```

```python
## Manhattan has the highest number of taxi pickups
```

```python
##★ SCATTER PLOT


sns.scatterplot(x="distance",y="fare",data=df)
plt.title("Distance vs Fare by Pickup Borough")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.show()
```

```python
##There is a positive correlation between distance and fare - as distance increases, fare also increases
```

```python
##★ Heatmap 
df.corr(numeric_only=True)
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True,fmt=".2f",cmap="Reds")
```

```python
##★ Pair Plot 

sns.pairplot(df,hue="pickup_borough")
plt.show()
```
