```python
#While loop & Control Statements:
#Number Gussing Game 

#1. set up the game
import random
secret_number =random.randint(1,10)

#2.Prompt the User
attempts=3
i=0
print("Guess your number 1 to 10")

#3.Implement the gussing logic

while i<attempts:
    number=input(f"Enter your guess: ")
    guess=int(number)
    if guess <1 or guess>10:
        print("Your guess is out of range.Please guess the number between 1 to 10")
        continue
    i+=1
    if guess==secret_number:
        print(f"Congrattulations!,You Guessed the correct answer")
        break
    elif guess<secret_number:
        print("Too low .Try again")

    else:
        print(f"Too high.Try again")

else:
            print("Better luck next time")
    
```

```python
#Function 

#BMI Calculator

def calculate_bmi(weight,height):
    bmi=float(weight/height**2)
    return bmi
    
weight=float(input("Enter your weight in kg: "))
height=float(input("Enter your height in meters: "))

     
result=calculate_bmi(weight,height)
print(f"Your BMI is:{result}")
```
